```python
import os
import pandas as pd

# --- DIRECTORIES ---
query_folder = "panels"         # the panels to evaluate
reference_folder = "atlas_panel" # where best matches should be searched
column_name = None              # set if CSVs have a named column
# ---------------------

def load_panels(folder):
    panels = {}
    for file in os.listdir(folder):
        if file.endswith(".csv"):
            df = pd.read_csv(os.path.join(folder, file))

            # automatically find gene column
            col = column_name if column_name else df.columns[0]

            panels[file] = set(df[col].astype(str))
    return panels

# Load query panels (to evaluate)
query_panels = load_panels(query_folder)

# Load reference panels (to compare against)
reference_panels = load_panels(reference_folder)

results = []

# For each panel in query folder, find best match in reference folder
for qfile, qgenes in query_panels.items():

    best_match = None
    best_overlap = -1
    best_size = None

    for rfile, rgenes in reference_panels.items():
        overlap = len(qgenes.intersection(rgenes))
        if overlap > best_overlap:
            best_overlap = overlap
            best_match = rfile
            best_size = len(rgenes)

    # percentages
    pct_query = best_overlap / len(qgenes) * 100 if len(qgenes) > 0 else 0
    pct_ref = best_overlap / best_size * 100 if best_size > 0 else 0

    # Save result
    results.append({
        "panel": qfile,
        "best_match": best_match,
        "overlap_size": best_overlap,
        "size_panel": len(qgenes),
        "size_best_match": best_size,
        "pct_overlap_query": pct_query,
    })

# Convert results to DataFrame
results_df = pd.DataFrame(results)
print("BEST PERCENTAGE POSSIBLE")
print(results_df)
```

```python
import os
import pandas as pd

hvg_folder = "HVG_panel"
de_folder = "test_panels_not_hvg_filtered"
column_name = None

def load_csv(path):
    df = pd.read_csv(path)
    col = column_name if column_name else df.columns[0]
    return set(df[col].astype(str))

def extract_size(filename):
    digits = ''.join([c for c in filename if c.isdigit()])
    return int(digits) if digits else None

# Build file maps but exclude files without sizes
hvg_files = {extract_size(f): f for f in os.listdir(hvg_folder)
             if f.endswith(".csv") and extract_size(f) is not None }

de_files = {extract_size(f): f for f in os.listdir(de_folder)
            if f.endswith(".csv") and extract_size(f) is not None and f.startswith("DE")}

results = []

matching_sizes = sorted(set(hvg_files.keys()) & set(de_files.keys()))

for size in matching_sizes:
    hvg_path = os.path.join(hvg_folder, hvg_files[size])
    de_path = os.path.join(de_folder, de_files[size])

    hvg_genes = load_csv(hvg_path)
    de_genes = load_csv(de_path)

    # Overlap
    overlap = len(hvg_genes.intersection(de_genes))
    pct_hvg = overlap / len(hvg_genes) * 100
    pct_de = overlap / len(de_genes) * 100

    # Unique genes
    hvg_unique = len(hvg_genes - de_genes)
    de_unique = len(de_genes - hvg_genes)
    union_total = len(hvg_genes.union(de_genes))

    results.append({
        "Panel Size": size,
        "HVG File": hvg_files[size],
        "DE File": de_files[size],
        "Overlap Size": overlap,
        "% overlap vs HVG": round(pct_hvg, 2),
        "% overlap vs DE": round(pct_de, 2),
        "HVG Unique": hvg_unique,
        "DE Unique": de_unique,
        "Total Unique (Union)": union_total
    })

df = pd.DataFrame(results)
print(df)


```

```python
import os
import pandas as pd

query_folder = "panels"              # lipid, aging, etc.
hvg_folder = "HVG_panel"
de_folder = "test_panels_not_hvg_filtered"
column_name = None

sizes = [16000, 8000, 4000, 2000, 1000]   # choose sizes to merge

def load_csv(path):
    df = pd.read_csv(path)
    col = column_name if column_name else df.columns[0]
    return set(df[col].astype(str))


# --------------------------------------
# Load biological query panels 
# --------------------------------------
query_panels = {
    f: load_csv(os.path.join(query_folder, f))
    for f in os.listdir(query_folder)
    if f.endswith(".csv")
}


# --------------------------------------
# Load Baseline
# --------------------------------------
baseline = load_csv(os.path.join(de_folder, "Atlas.csv"))

reference_panels = {"Baseline": baseline}

# Track merged unique sizes
merged_unique_table = []

# --------------------------------------
# BUILD MERGED PANELS = HVG ∪ DE
# --------------------------------------
for size in sizes:
    hvg_file = f"HVG_panel_{size}.csv"
    de_file  = f"DE_panel_top{size}.csv"

    if os.path.exists(os.path.join(hvg_folder, hvg_file)) and \
       os.path.exists(os.path.join(de_folder, de_file)):

        hvg_genes = load_csv(os.path.join(hvg_folder, hvg_file))
        de_genes  = load_csv(os.path.join(de_folder, de_file))

        merged = hvg_genes.union(de_genes)

        reference_panels[f"MERGED_{size}"] = merged

        # STORE UNIQUE COUNT
        merged_unique_table.append({
            "Merged Panel": f"MERGED_{size}",
            "HVG size": len(hvg_genes),
            "DE size": len(de_genes),
            "Union size (Unique genes)": len(merged)
        })


# --------------------------------------
# MAIN TABLE: Overlap of each biological panel vs merged/baseline
# --------------------------------------
rows = []

for panel_name, genes in query_panels.items():
    row = {"Panel": panel_name, "Panel Size": len(genes)}

    for ref_name, ref_genes in reference_panels.items():
        overlap = len(genes & ref_genes)
        pct = overlap / len(genes) * 100
        row[ref_name] = f"{overlap} ({pct:.2f}%)"

    rows.append(row)

df = pd.DataFrame(rows)

# reorder columns
cols = ["Panel", "Baseline"] + [f"MERGED_{s}" for s in sizes] + ["Panel Size"]
df = df[cols]


# --------------------------------------
# SECOND TABLE — UNIQUE GENES IN MERGED PANELS
# --------------------------------------
df_unique = pd.DataFrame(merged_unique_table)

# sort by numeric size
df_unique["size"] = df_unique["Merged Panel"].str.extract(r"(\d+)").astype(int)
df_unique = df_unique.sort_values("size").drop(columns=["size"])


# --------------------------------------
# PRINT BOTH TABLES
# --------------------------------------
print("\n===== MAIN PANEL OVERLAP TABLE =====\n")
print(df)

print("\n===== UNIQUE GENES PER MERGED PANEL =====\n")
print(df_unique)


# Save to disk
df.to_csv("panel_overlap_with_merged.csv", index=False)
df_unique.to_csv("merged_panel_unique_gene_counts.csv", index=False)

print("\nSaved outputs:")
print(" - panel_overlap_with_merged.csv")
print(" - merged_panel_unique_gene_counts.csv")

```

```python
import os
import pandas as pd
import re

hvg_folder = "HVG_panel"
panel_folder = "test_panels_not_hvg_filtered"
column_name = None

sizes = [16000, 8000, 4000, 2000, 1000]

def load_csv(path):
    df = pd.read_csv(path)
    col = df.columns[0] if column_name is None else column_name
    return set(df[col].astype(str))

def extract_size(filename):
    m = re.findall(r"\d+", filename)
    return int(m[-1]) if m else None

# Load HVG, DE, PCA for each size
hvg_panels = {}
de_panels = {}
pca_panels = {}

for size in sizes:
    hf = f"HVG_panel_{size}.csv"
    df = f"DE_panel_top{size}.csv"
    pf = f"PCA_panel_top{size}.csv"

    if os.path.exists(os.path.join(hvg_folder, hf)):
        hvg_panels[size] = load_csv(os.path.join(hvg_folder, hf))

    if os.path.exists(os.path.join(panel_folder, df)):
        de_panels[size] = load_csv(os.path.join(panel_folder, df))

    if os.path.exists(os.path.join(panel_folder, pf)):
        pca_panels[size] = load_csv(os.path.join(panel_folder, pf))

valid_sizes = sorted(set(hvg_panels) & set(de_panels) & set(pca_panels))

results = []

for size in valid_sizes:
    HVG = hvg_panels[size]
    DE  = de_panels[size]
    PCA = pca_panels[size]

    # pairwise overlaps
    overlap_hvg_de = len(HVG & DE)
    overlap_hvg_pca = len(HVG & PCA)
    overlap_de_pca = len(DE & PCA)
    overlap_all3 = len(HVG & DE & PCA)

    # unions
    union_hvg_de = len(HVG | DE)
    union_hvg_pca = len(HVG | PCA)
    union_de_pca = len(DE | PCA)
    union_all3 = len(HVG | DE | PCA)

    # unique parts
    unique_hvg_in_hvg_de = len(HVG - DE)
    unique_de_in_hvg_de = len(DE - HVG)

    unique_hvg_in_hvg_pca = len(HVG - PCA)
    unique_pca_in_hvg_pca = len(PCA - HVG)

    unique_de_in_de_pca = len(DE - PCA)
    unique_pca_in_de_pca = len(PCA - DE)

    unique_hvg_all3 = len(HVG - DE - PCA)
    unique_de_all3 = len(DE - HVG - PCA)
    unique_pca_all3 = len(PCA - HVG - DE)

    results.append({
        "Size": size,

        # overlaps
        "Overlap_HVG_DE": overlap_hvg_de,
        "Overlap_HVG_PCA": overlap_hvg_pca,
        "Overlap_DE_PCA": overlap_de_pca,
        "Overlap_ALL3": overlap_all3,

        # unions
        "Union_HVG_DE": union_hvg_de,
        "Union_HVG_PCA": union_hvg_pca,
        "Union_DE_PCA": union_de_pca,
        "Union_ALL3": union_all3,

        # unique parts
        "Unique_HVG_in_HVG_DE": unique_hvg_in_hvg_de,
        "Unique_DE_in_HVG_DE": unique_de_in_hvg_de,

        "Unique_HVG_in_HVG_PCA": unique_hvg_in_hvg_pca,
        "Unique_PCA_in_HVG_PCA": unique_pca_in_hvg_pca,

        "Unique_DE_in_DE_PCA": unique_de_in_de_pca,
        "Unique_PCA_in_DE_PCA": unique_pca_in_de_pca,

        "Unique_HVG_ALL3": unique_hvg_all3,
        "Unique_DE_ALL3": unique_de_all3,
        "Unique_PCA_ALL3": unique_pca_all3
    })

df = pd.DataFrame(results)
print(df)


```

```python
import os
import pandas as pd
import re

query_folder = "panels"              
hvg_folder = "HVG_panel"
de_folder = "test_panels_not_hvg_filtered"
pca_folder = "test_panels_not_hvg_filtered"
column_name = None

sizes = [8000, 4000, 2000, 1000]

def load_csv(path):
    df = pd.read_csv(path)
    col = df.columns[0] if column_name is None else column_name
    return set(df[col].astype(str))

def extract_percent(cell):
    """Extracts percentage from 'NN (XX.XX%)'."""
    if isinstance(cell, str) and "(" in cell and "%" in cell:
        return float(cell.split("(")[1].split("%")[0])
    return None


# -----------------------------------
# Load biological panels
# -----------------------------------
query_panels = {
    f: load_csv(os.path.join(query_folder, f))
    for f in os.listdir(query_folder)
    if f.endswith(".csv")
}

# -----------------------------------
# Load HVG, DE, PCA sets
# -----------------------------------
hvg = {}
de = {}
pca = {}

for size in sizes:
    hf = f"HVG_panel_{size}.csv"
    df = f"DE_panel_top{size}.csv"
    pf = f"PCA_panel_top{size}.csv"

    if os.path.exists(os.path.join(hvg_folder, hf)):
        hvg[size] = load_csv(os.path.join(hvg_folder, hf))

    if os.path.exists(os.path.join(de_folder, df)):
        de[size] = load_csv(os.path.join(de_folder, df))

    if os.path.exists(os.path.join(pca_folder, pf)):
        pca[size] = load_csv(os.path.join(pca_folder, pf))


# -----------------------------------
# Compute overlaps (values + percentages)
# -----------------------------------
rows = []

for panel_name, genes in query_panels.items():
    panel_size = len(genes)
    row = {"Panel": panel_name, "Panel Size": panel_size}

    for size in sizes:
        if size not in hvg or size not in de or size not in pca:
            continue

        h = hvg[size]
        d = de[size]
        p = pca[size]

        # unions
        h_d   = h | d
        h_p   = h | p
        d_p   = d | p
        h_d_p = h | d | p

        # overlaps
        ov1 = len(genes & h_d)
        ov2 = len(genes & h_p)
        ov3 = len(genes & d_p)
        ov4 = len(genes & h_d_p)

        # formatted entries
        row[f"HVG+DE_{size}"] = f"{ov1} ({ov1/panel_size*100:.2f}%)"
        row[f"HVG+PCA_{size}"] = f"{ov2} ({ov2/panel_size*100:.2f}%)"
        row[f"DE+PCA_{size}"] = f"{ov3} ({ov3/panel_size*100:.2f}%)"
        row[f"HVG+DE+PCA_{size}"] = f"{ov4} ({ov4/panel_size*100:.2f}%)"

    rows.append(row)

df = pd.DataFrame(rows)

# -----------------------------------
# ADD MEAN PERCENTAGE ROW
# -----------------------------------
mean_row = {"Panel": "MEAN_PERCENT", "Panel Size": ""}

for size in sizes:
    for key_prefix in ["HVG+DE", "HVG+PCA", "DE+PCA", "HVG+DE+PCA"]:
        col = f"{key_prefix}_{size}"

        if col in df.columns:
            percentages = df[col].apply(extract_percent)
            mean_pct = percentages.mean()
            mean_row[col] = f"{mean_pct:.2f}%"

df = pd.concat([df, pd.DataFrame([mean_row])], ignore_index=True)



# Save
df.to_csv("panel_overlap_simple_with_percent.csv", index=False)
print(df)
print("\nSaved to panel_overlap_simple_with_percent.csv")

```

```python
import numpy as np
import pandas as pd
from numba import njit, prange
from tqdm import tqdm

# ============================================================
#      A-E-S (Efraimidis & Spirakis) – Weighted W/O Replacement
# ============================================================

@njit
def aes_weighted_sample_no_replacement(weights, k):
    """
    Weighted sampling without replacement using A-ES algorithm:
    key_i = U_i^(1 / w_i)
    Pick top-k keys.
    """
    n = weights.size
    keys = np.empty(n)

    for i in range(n):
        u = np.random.random()
        keys[i] = u ** (1.0 / weights[i])

    # argpartition più veloce di sort
    idx = np.argpartition(keys, -k)[-k:]
    return idx


# ============================================================
#            Monte Carlo veloce con Numba
# ============================================================

@njit(parallel=True)
def run_simulations(weights, real_idx, threshold, N_SAMPLE, N_SIM):
    count = 0

    # per ricerca veloce: converti pannello reale in set booleano O(1)
    max_idx = weights.size
    mask = np.zeros(max_idx, dtype=np.uint8)
    for r in real_idx:
        mask[r] = 1

    for s in prange(N_SIM):
        chosen = aes_weighted_sample_no_replacement(weights, N_SAMPLE)

        # conta overlap
        overlap = 0
        for c in chosen:
            if mask[c] == 1:
                overlap += 1

        if overlap >= threshold:
            count += 1

    return count


# ============================================================
#                ESEMPIO COMPLETO DI UTILIZZO
# ============================================================

# --- carica file probabilità ---
df = pd.read_csv("score_panels/combined_panel_sorted.csv")
genes = df.iloc[:,0].values
weights = df.iloc[:,1].values.astype(np.float64)
weights = weights / weights.sum()  # normalizza (essenziale)

# --- carica pannello reale ---
panel_real = pd.read_csv("panels/Allen_Zeng.csv", header=None).iloc[:,0].values
panel_real = list(set(panel_real))  # rimuove duplicati

# mappa gene → indice
gene_to_idx = {g: i for i, g in enumerate(genes)}

# converte pannello reale in indici nel vettore pesi
real_idx = []
for g in panel_real:
    if g in gene_to_idx:
        real_idx.append(gene_to_idx[g])
real_idx = np.array(real_idx)

# parametri
N_SAMPLE = 1000
OVERLAP_FRAC = 0.10
threshold = int(np.ceil(len(real_idx) * OVERLAP_FRAC))
N_SIM = 100000   # cambia quanto vuoi

print("Geni nel pannello reale trovati:", len(real_idx))
print("Soglia overlap:", threshold)

# ----------- RUN -------------
print("\nCompilazione Numba… (solo la 1ª volta è lenta)")
count = run_simulations(weights, real_idx, threshold, N_SAMPLE, N_SIM)

prob = count / N_SIM
print("\n================ RISULTATO ================")
print(f"P(overlap ≥ {OVERLAP_FRAC*100:.1f}%) ≈ {prob:.6e}")
print("==========================================")
```

```python
import os
import numpy as np
import pandas as pd

PROB_FILE = "score_panels/combined_panel_sorted_max.csv"  # gene_id, prob/weight
PANEL_FOLDER = "panels"

TRIM_FRAC = 0.05            # trim 5% top e 5% bottom
DROP_TOP_K_LIST = [1, 5, 10, 20, 40, 60]
STABILITY_DROP_FRACS = [0.0, 0.05, 0.10, 0.20, 0.40, 0.60]  # togli 5/10/20% geni (più "outlier-like")
SEED = 0

rng = np.random.default_rng(SEED)

# -----------------------
# carica pesi globali
# -----------------------
df = pd.read_csv(PROB_FILE)
gene_col = df.columns[0]
w_col = df.columns[1]

genes = df[gene_col].astype(str).values
w = df[w_col].astype(float).values
w = w / w.sum()
N = len(w)

gene_to_w = dict(zip(genes, w))

print(f"N geni nel file pesi: {N}")
print(f"Somma pesi totale: {w.sum():.6f}\n")

def panel_weights(panel_genes):
    # pesi dei geni del pannello (solo quelli presenti nel file pesi)
    arr = np.array([gene_to_w[g] for g in panel_genes if g in gene_to_w], dtype=float)
    return arr

def trimmed_sum(x, trim_frac):
    if x.size == 0:
        return 0.0
    x_sorted = np.sort(x)
    t = int(np.floor(trim_frac * x_sorted.size))
    if 2*t >= x_sorted.size:
        return x_sorted.mean() * x_sorted.size  # fallback
    return x_sorted[t: x_sorted.size - t].sum()

def drop_top_k_sum(x, k):
    if x.size == 0:
        return 0.0
    if k <= 0:
        return x.sum()
    if k >= x.size:
        return 0.0
    x_sorted = np.sort(x)
    return x_sorted[: x_sorted.size - k].sum()

print("=== PANEL ROBUST WEIGHT SCORES (no sampling) ===\n")

for fname in sorted(os.listdir(PANEL_FOLDER)):
    if not fname.endswith(".csv"):
        continue

    pdf = pd.read_csv(os.path.join(PANEL_FOLDER, fname))
    col = pdf.columns[0]
    panel_genes = pdf[col].astype(str).values
    panel_genes = set(panel_genes)

    x = panel_weights(panel_genes)
    k = x.size
    if k == 0:
        continue

    # baseline uniforme per somma-pesi raw
    uniform_sum = k / N

    raw = x.sum()
    trim = trimmed_sum(x, TRIM_FRAC)
    med = float(np.median(x))

    print(f"PANEL: {fname}")
    print(f"  |P| (matchati nel file pesi) = {k}")
    print(f"  Uniform baseline (|P|/N)     = {uniform_sum:.6e}")
    print(f"  Raw sum(weights)             = {raw:.6e}   (enr={raw/uniform_sum:.3f}x)")
    print(f"  Trimmed sum ({TRIM_FRAC*100:.1f}% each tail) = {trim:.6e}   (enr={trim/uniform_sum:.3f}x)")
    print(f"  Median weight                = {med:.6e}")

    # drop-top-k (outlier removal deterministica)
    for dk in DROP_TOP_K_LIST:
        val = drop_top_k_sum(x, dk)
        print(f"  Drop top {dk:2d} weights sum      = {val:.6e}   (enr={val/uniform_sum:.3f}x)")

    # stabilità: quanto cambia lo score se togli una frazione di geni dal pannello
    # Qui faccio una cosa COERENTE con quello che chiedi: tolgo geni (non dal genoma),
    # e guardo la distribuzione dello score. Per togliere "outlier" in modo mirato,
    # tolgo i più grandi (worst-case outlier sensitivity).
    x_sorted = np.sort(x)  # crescente
    print("  Robustness by removing largest weights:")
    for frac in STABILITY_DROP_FRACS:
        r = int(np.floor(frac * k))
        if r >= k:
            continue
        # rimuovi i r pesi più grandi
        val = x_sorted[: k - r].sum()
        print(f"    remove top {frac*100:4.0f}% (={r:3d} genes): sum = {val:.6e} (enr={val/uniform_sum:.3f}x)")

    print("-" * 75)

```

```python
import os
import pandas as pd
import numpy as np

RANK_FILE = "score_panels/combined_panel_sorted_max.csv"
PANEL_FOLDER = "panels"

# percentili da analizzare
PCTS = [5, 10, 25, 50]

# --------------------------------------------------
# 1. Carica ranking globale
# --------------------------------------------------
df = pd.read_csv(RANK_FILE)
gene_col = df.columns[0]

ranked_genes = df[gene_col].astype(str).values
N = len(ranked_genes)

gene_to_rank = {g: i + 1 for i, g in enumerate(ranked_genes)}

# soglie di rank
rank_thresholds = {
    p: int(np.floor(p / 100 * N))
    for p in PCTS
}

print(f"Totale geni nel ranking: {N}")
print("Rank thresholds:")
for p, r in rank_thresholds.items():
    print(f"  Top {p:2d}%  -> rank ≤ {r}")
print()

# --------------------------------------------------
# 2. Analisi pannelli
# --------------------------------------------------
print("=========== PANEL TOP-PERCENT OVERLAP ===========\n")

for fname in sorted(os.listdir(PANEL_FOLDER)):
    if not fname.endswith(".csv"):
        continue

    pdf = pd.read_csv(os.path.join(PANEL_FOLDER, fname))
    col = pdf.columns[0]
    panel_genes = set(pdf[col].astype(str))

    panel_ranks = [
        gene_to_rank[g] for g in panel_genes
        if g in gene_to_rank
    ]

    k = len(panel_ranks)
    if k == 0:
        continue

    print(f"PANEL: {fname}")
    print(f"  |P| = {k}")

    for p in PCTS:
        thr = rank_thresholds[p]
        cnt = sum(r <= thr for r in panel_ranks)
        frac = cnt / k * 100
        print(f"  Top {p:2d}%: {cnt:4d} genes  ({frac:5.1f}%)")

    print("-" * 55)

print("\n=================================================\n")
```
